### Teraz importujemy biblioteki oraz wprowadzamny dane za pomocą biblioteki pandas

import pandas as pd

# Dane symulowane
data = {
    'Miesiąc': pd.date_range(start='2023-01-01', periods=12, freq='M'),
    'Sprzedaż': [870, 1050, 1200, 1150, 1300, 1700, 1400, 1800, 1300, 1600, 1150, 1600],
    'Promocja (%)': [0, 25, 5, 15, 0, 10, 15, 20, 10, 0, 5, 30],
    'Cena (PLN)': [5.49, 5.79, 5.99, 6.19, 6.29, 6.49, 6.39, 6.59, 6.49, 6.79, 6.69, 6.99]

df = pd.DataFrame(data)
print(df)


### Rysujemy wykresy

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Wizualizacja sprzedaży
plt.figure(figsize=(10, 6))
sns.lineplot(x='Miesiąc', y='Sprzedaż', data=df, marker='o', label='Sprzedaż')
plt.title('Sprzedaż miesięczna produktów')
plt.xlabel('Miesiąc')
plt.ylabel('Liczba sprzedanych sztuk')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Wizualizacja sprzedaży względem promocji
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Promocja (%)', y='Sprzedaż', data=df, hue='Cena (PLN)', palette='viridis', s=100)
plt.title('Sprzedaż w zależności od promocji i ceny')
plt.xlabel('Promocja (%)')
plt.ylabel('Sprzedaż')
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
# Wizualizacja sprzedaży względem ceny
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Cena (PLN)', y='Sprzedaż', data=df, hue='Promocja (%)', palette='coolwarm', s=100)
plt.title('Sprzedaż w zależności od ceny i promocji')
plt.xlabel('Cena (PLN)')
plt.ylabel('Sprzedaż')
plt.grid(True)
plt.tight_layout()
plt.show()
```

# Wnioski:

- Najwyższą sprzedaż można zauważyć w miesiącach: lipiec, sierpień,wrzesień ze względu na fakt że arbuz jest sprzedawany najwięcej w okresie letnim. 
- Istnieje pewna zależność pomiędzy wynikami sprzedażą a promocjami. W miesiącach, w których promocja wynosiła 20,25 i 30 % sprzedaż utrzymała się na bardzo wysokim poziomie.
- Wraz ze wzrostem cen wyniki sprzedaży wzrastały ze względu na istniejące promocje.Czyli jeśli będziemy tworzyć promocje jednocześnie podwyższając cenę, sprzedaż wzrośnie lub utrzyma się na podobnym poziomie
- Wyższy rabat zwykle wiąże się z większą sprzedażą, szczególnie przy umiarkowanej cenie produktu;
- Wysoka cena, niepoparta promocją (np. październik), może skutkować spadkiem sprzedaży mimo wcześniejszych wzrostów.
